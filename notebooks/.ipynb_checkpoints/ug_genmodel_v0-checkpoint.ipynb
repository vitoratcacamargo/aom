## Generative Model

```julia
const NOUNS = UGmodel.LEXICON.nouns
const VERBS_TRANS = UGmodel.LEXICON.verbs_trans
const VERBS_INTR = UGmodel.LEXICON.verbs_intr
const ADJECTIVES = UGmodel.LEXICON.adjectives
const WH_WORD = UGmodel.LEXICON.wh_word
```

```julia
# Creating fixed size vocabulary
const VOCAB = unique(vcat(["PAD", WH_WORD], NOUNS, VERBS_TRANS, VERBS_INTR, ADJECTIVES))
const VOCAB_SIZE = length(VOCAB)
const WORD_TO_IDX = Dict(w => i for (i,w) in enumerate(VOCAB))
const MAX_SENT_LEN = 6
```

```julia
# Noise constant to account for misunderstandings, slip of the tongue etc.
# Generates ungrammatical sentences
const NOISE_PROB = 0.1
```

```julia
function emission_probs(target_idx::Int; V::Int = VOCAB_SIZE, epsilon::Float64 = 0.05)
    @assert 0 < epsilon < 1
    probs = fill(epsilon / (V-1), V)
    probs[target_idx] = 1.0 -epsilon

    return probs

end
```

```julia
@gen function corpus_model(num_sentences::Int)
    # === Universal Grammar paramaters (theta) ===
    # Head direction: 1 = Verb-Object (VO), 0 = Object-Verb(OV)
    hd = @trace(bernoulli(0.5), :head_direction)

    # Subject drop: 1 = subject may be dropped, 0 = subject always present
    sd = @trace(bernoulli(0.5), :subject_drop)

    # Adjective order: 1 = Adjective-Noun (AN), 0 = Noun-Adjective (NA)
    ao = @trace(bernoulli(0.5), :adj_order)

    # Wh-fronting in question: 1 = wh-word at front, 0 = in-situ
    wh = @trace(bernoulli(0.5), :wh_fronting)
    
    sentences = Vector{Vector{String}}(undef, num_sentences)

    for i in 1:num_sentences
        # Utterance type: 1 = transitive declarative, 2 = intransitive, 3 = transitive question
        utt_type = @trace(categorical([0.4, 0.3, 0.3]), (:utt_type, i))

        # Sample lexical items
        subj_idx = @trace(categorical(fill(1/length(NOUNS), length(NOUNS))), (:subj, i))
        obj_idx = @trace(categorical(fill(1/length(NOUNS), length(NOUNS))), (:obj, i))
        adj_idx = @trace(categorical(fill(1/length(ADJECTIVES), length(ADJECTIVES))), (:adj, i))

        if utt_type == 1 || utt_type == 3
            v_idx = @trace(categorical(fill(1/length(VERBS_TRANS), length(VERBS_TRANS))), (:v_trans, i))
            verb = VERBS_TRANS[v_idx]
        else
            v_idx = @trace(categorical(fill(1/length(VERBS_INTR), length(VERBS_INTR))), (:v_intr, i))
            verb = VERBS_INTR[v_idx]
        end

        subj = NOUNS[subj_idx]
        obj = NOUNS[obj_idx]
        adj = ADJECTIVES[adj_idx]

        # Add adjective order error
        ao_slip = @trace(bernoulli(NOISE_PROB), (:ao_slip, i))
        effective_ao = ao_slip ? !ao : ao
        
        # Build NP for subject and object, respecting adjective order and taking error into account
        function make_np(noun::String, adj::String, order_rule::Bool)
            if order_rule == 1
                return [adj, noun] # Adj N
            else
                return [noun, adj] # N Adj
            end
        end

        subj_np = make_np(subj, adj, effective_ao)
        obj_np = make_np(obj, adj, effective_ao)

        # Add head direction error
        hd_slip = @trace(bernoulli(NOISE_PROB), (:hd_slip, i))
        effective_hd = hd_slip ? !hd : hd

        # Base clause word order for transitive
        vp = Vector{String}()
        if utt_type == 1 || utt_type == 3
            # Verb-Object vs. Object-Verb in the VP
            vp = (effective_hd == 1) ? vcat([verb], obj_np) : vcat(obj_np, [verb])
        else
            # Intransitive: just the verb
            vp = [verb]
        end

        # Add subject drop error
        sd_slip = @trace(bernoulli(NOISE_PROB), (:sd_slip, i))
        effective_sd = sd_slip ? !sd : sd

        # Drop subject if SD = 1
        drop_subj = (effective_sd == 1) ? @trace(bernoulli(0.5), (:drop_subj, i)) : 0

        clause = Vector{String}()

        if drop_subj == 0
            append!(clause, subj_np)
        end
        
        append!(clause, vp)

        # Handle questions
        if utt_type == 3
            wh_slip = @trace(bernoulli(NOISE_PROB), (:wh_slip, i))
            effective_wh = wh_slip ? !wh : wh
            
            if effective_wh == 1
                # wh-fronting
                clause = vcat([WH_WORD], clause)
            else
                # wh in-situ (append at the end)
                push!(clause, WH_WORD)
            end
        end

        sentences[i] = clause

        for pos in 1:MAX_SENT_LEN
            word = pos <= length(clause) ? clause[pos] : "PAD"
            vocab_idx = WORD_TO_IDX[word]
            probs = emission_probs(vocab_idx; V=VOCAB_SIZE, epsilon = NOISE_PROB * 0.5)
            @trace(categorical(probs), (:token, i, pos))
        end
    end

    return sentences
end
```

```julia
trace = simulate(corpus_model, (10,))
sentences = get_retval(trace)
theta = (
    hd = trace[:head_direction],
    sd = trace[:subject_drop],
    ao = trace[:adj_order],
    wh = trace[:wh_fronting]
)
theta
```

## Neural Network

```julia
# Neural net params
const NUM_SENTS_TRAIN = 10
const INPUT_DIM = VOCAB_SIZE * MAX_SENT_LEN * NUM_SENTS_TRAIN
H = 128
```

```julia
mutable struct NeuralNetParams
    W1::Matrix{Float64}  
    b1::Vector{Float64}  
    W2::Matrix{Float64}  
    b2::Vector{Float64}  
end
```

### Encoding the sentences

```julia
# Convert sentences into a flattened one hot encoded vector
function encode_corpus(sentences::Vector{Vector{String}})
    encoded = zeros(Float64, INPUT_DIM)

    cursor = 0
    for i in 1:NUM_SENTS_TRAIN
        if i > length(sentences) break end

        sent = sentences[i]
        for j in 1:MAX_SENT_LEN
            word = j <= length(sent) ? sent[j] : "PAD"

            vocab_idx = WORD_TO_IDX[word]

            encoded[cursor * VOCAB_SIZE + vocab_idx] = 1.0
            cursor += 1
        end
    end

    return encoded
end
```

### Creating the innate solver neural network (Variational Approximation)

```julia
σ(x) = tanh.(x)
sigmoid(x) = 1.0 ./ (1.0 .+ exp.(-x))

@gen function innate_solver(input_vector::Vector{Float64})
    @param W1::Matrix{Float64}
    @param b1::Vector{Float64}
    @param W2::Matrix{Float64}
    @param b2::Vector{Float64}

    hidden_layer = σ(W1 * input_vector + b1)
    output = W2 * hidden_layer + b2

    probs = sigmoid(output)

    for i in 1:4
        @trace(Gen.bernoulli(probs[i]), (:param, i))
    end
    
    return probs
end

init_weight(out, inc) = randn(out, inc) * 0.01
init_param!(innate_solver, :W1, init_weight(H, INPUT_DIM))
init_param!(innate_solver, :b1, zeros(H))
init_param!(innate_solver, :W2, init_weight(4, H))
init_param!(innate_solver, :b2, zeros(4))
```

### Generating synthetic languages from the GM

```julia
function supervised_training_batch_generator()
    tr = simulate(corpus_model, (NUM_SENTS_TRAIN,))
    sentences = get_retval(tr)

    input_vec = encode_corpus(sentences)

    constraints = Gen.choicemap()
    constraints[(:param, 1)] = tr[:head_direction]
    constraints[(:param, 2)] = tr[:subject_drop]
    constraints[(:param, 3)] = tr[:adj_order]
    constraints[(:param, 4)] = tr[:wh_fronting]

    return ((input_vec,), constraints)
end
```

### Training the innate solver

```julia
update = Gen.ParamUpdate(Gen.FixedStepGradientDescent(0.001), innate_solver)

NUM_EPOCHS = 500
EPOCH_SIZE = 50

TRAIN = true

if TRAIN
    losses = Float64[]
    
    for epoch in 1:NUM_EPOCHS
        epoch_loss = 0.0
        
        for i in 1:EPOCH_SIZE
            (args, constraints) = supervised_training_batch_generator()
            (trace, weight) = generate(innate_solver, args, constraints)
            
            Gen.accumulate_param_gradients!(trace)
            
            epoch_loss += -get_score(trace)
        end
        
        Gen.apply!(update)
        
        avg_loss = epoch_loss / EPOCH_SIZE
        push!(losses, avg_loss)
        
        if epoch == 1 || epoch % 10 == 0
            println("Epoch $epoch: avg loss = $(round(avg_loss, digits=4))")
        end
    end

    plot(losses, label = "Loss (Neg LogLik)", xlabel = "Epochs", title = "Training Innate Solver")
end
```

## Comparison Innate Solver vs Naive Learner

```julia
function predict_nn(input_vec::Vector{Float64})
    W1 = get_param(innate_solver, :W1)
    b1 = get_param(innate_solver, :b1)
    W2 = get_param(innate_solver, :W2)
    b2 = get_param(innate_solver, :b2)

    hidden = σ(W1 * input_vec + b1)
    output = W2 * hidden + b2
    probs = sigmoid(output)

    preds = (
        hd = probs[1] > 0.5,
        sd = probs[2] > 0.5,
        ao = probs[3] > 0.5,
        wh = probs[4] > 0.5,
    )

    return preds, probs
end
```

```julia
# MH update on the 4 parameters
function mh_step(tr)
    tr, _ = mh(tr, select(:head_direction))
    tr, _ = mh(tr, select(:subject_drop))
    tr, _ = mh(tr, select(:adj_order))
    tr, _ = mh(tr, select(:wh_fronting))
    return tr
end
```

```julia
function is_correct(tr)
    return tr[:head_direction] == target_theta.hd &&
           tr[:subject_drop] == target_theta.sd &&
           tr[:adj_order] == target_theta.ao &&
           tr[:wh_fronting] == target_theta.wh
end
```

```julia
# Define the ground truth constraints. This is set to be somewhat English-like
target_theta = (hd=true, sd=false, ao=true, wh=true)

target_constraints = Gen.choicemap()
target_constraints[:head_direction] = target_theta.hd
target_constraints[:subject_drop] = target_theta.sd
target_constraints[:adj_order] = target_theta.ao
target_constraints[:wh_fronting] = target_theta.wh

# Generate data with the constraints
tr_target, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), target_constraints)
observed_sentences = get_retval(tr_target)
observed_input_vec = encode_corpus(observed_sentences)

# Create the observation constraints
# Note that this constrains the trace to match the observed tokens to make inference more tractable
# Since we're observing the capability of inferring the underlying parameters, guessing the tokens is not relevant to the task
obs_map = Gen.choicemap()
for i in 1:NUM_SENTS_TRAIN
    obs_map[(:utt_type, i)] = tr_target[(:utt_type, i)]
    obs_map[(:subj, i)] = tr_target[(:subj, i)]
    obs_map[(:obj, i)] = tr_target[(:obj, i)]
    obs_map[(:adj, i)] = tr_target[(:adj, i)]

    if tr_target[(:utt_type, i)] == 1 || tr_target[(:utt_type, i)] == 3
         obs_map[(:v_trans, i)] = tr_target[(:v_trans, i)]
    else
         obs_map[(:v_intr, i)] = tr_target[(:v_intr, i)]
    end

    for j in 1:MAX_SENT_LEN
        obs_map[(:token, i, j)] = tr_target[(:token, i, j)]
    end
end
```

### Naive Learner with random initialization

```julia
naive_trace, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), obs_map)

naive_acc = Float64[]
push!(naive_acc, is_correct(naive_trace) ? 1.0 : 0.0)

for i in 1:500
    global naive_trace = mh_step(naive_trace)
    push!(naive_acc, is_correct(naive_trace) ? 1.0 : 0.0)
end
```

### Guided Learner

```julia
# Get the guided guess from the neural network
nn_preds, nn_probs = predict_nn(observed_input_vec)

# Get the initial observation constraints
guided_init_map = deepcopy(obs_map)

# Add the neural network predictions as initial values
guided_init_map[:head_direction] = nn_preds.hd
guided_init_map[:subject_drop] = nn_preds.sd
guided_init_map[:adj_order] = nn_preds.ao
guided_init_map[:wh_fronting] = nn_preds.wh

guided_trace, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), guided_init_map)

guided_acc = Float64[]
push!(guided_acc, is_correct(guided_trace) ? 1.0 : 0.0)

for i in 1:500
    global guided_trace = mh_step(guided_trace)
    push!(guided_acc, is_correct(guided_trace) ? 1.0 : 0.0)
end
```

## Results Visualization

```julia
function run_comparison_trial()
    # Naive learner
    naive_trace, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), obs_map)
    naive_params, _ = track_parameters(naive_trace, 500)
    _, _, naive_conv = compute_convergence_metrics(naive_params, target_theta)
    
    # Guided learner
    nn_preds, _ = predict_nn(observed_input_vec)
    guided_init_map = deepcopy(obs_map)
    guided_init_map[:head_direction] = nn_preds.hd
    guided_init_map[:subject_drop] = nn_preds.sd
    guided_init_map[:adj_order] = nn_preds.ao
    guided_init_map[:wh_fronting] = nn_preds.wh
    guided_trace, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), guided_init_map)
    guided_params, _ = track_parameters(guided_trace, 500)
    _, _, guided_conv = compute_convergence_metrics(guided_params, target_theta)
    
    return naive_conv, guided_conv
end

# Run multiple trials
n_trials = 20
println("Running $n_trials trials...")
results = [run_comparison_trial() for _ in 1:n_trials]
naive_convs = [r[1] for r in results]
guided_convs = [r[2] for r in results]

p = plot(layout=(1,2), size=(1200, 400))

# Box plot
boxplot!(p[1], 
         ["Naive", "Guided"],
         [naive_convs, guided_convs],
         title="Convergence Speed Distribution",
         ylabel="Steps to Convergence",
         legend=false,
         color=[:red :blue])

# Histogram
histogram!(p[2], naive_convs,
          label="Naive",
          alpha=0.6,
          color=:red,
          bins=20,
          title="Convergence Speed Distribution",
          xlabel="Steps to Convergence",
          ylabel="Frequency")
histogram!(p[2], guided_convs,
          label="Guided",
          alpha=0.6,
          color=:blue,
          bins=20)

display(p)
println("\nStatistics:")
println("Naive - Mean: $(round(mean(naive_convs), digits=1)), Std: $(round(std(naive_convs), digits=1))")
println("Guided - Mean: $(round(mean(guided_convs), digits=1)), Std: $(round(std(guided_convs), digits=1))")
println("Average speedup: $(round(mean(naive_convs) / mean(guided_convs), digits=2))x")
```

```julia
## Visualizing Generated Sentences
"""
Visualize the structure of generated sentences to understand 
how UG parameters manifest in surface forms.
"""

function visualize_sentences(sentences, theta; max_display=8)
    println("=" ^ 60)
    println("Generated Sentences with UG Parameters:")
    println("  Head Direction: $(theta.hd ? "VO (Verb-Object)" : "OV (Object-Verb)")")
    println("  Subject Drop: $(theta.sd ? "Allowed" : "Required")")
    println("  Adjective Order: $(theta.ao ? "AN (Adj-Noun)" : "NA (Noun-Adj)")")
    println("  Wh-Fronting: $(theta.wh ? "Fronted" : "In-situ")")
    println("=" ^ 60)
    
    for (i, sent) in enumerate(sentences[1:min(length(sentences), max_display)])
        println("  $i. $(join(sent, " "))")
    end
    println()
end

# Demonstrate different language types
println("\n### Example Languages from Different UG Settings ###\n")

language_configs = [
    (hd=true, sd=false, ao=true, wh=true, name="English-like"),
    (hd=false, sd=true, ao=false, wh=false, name="Japanese-like"),
    (hd=true, sd=true, ao=true, wh=true, name="Spanish-like"),
    (hd=false, sd=false, ao=true, wh=false, name="German-like (partial)")
]

for config in language_configs
    constraints = Gen.choicemap()
    constraints[:head_direction] = config.hd
    constraints[:subject_drop] = config.sd
    constraints[:adj_order] = config.ao
    constraints[:wh_fronting] = config.wh
    
    tr, _ = generate(corpus_model, (5,), constraints)
    sents = get_retval(tr)
    theta = (hd=config.hd, sd=config.sd, ao=config.ao, wh=config.wh)
    
    println("\n### $(config.name) ###")
    visualize_sentences(sents, theta; max_display=5)
end
```

```julia
## Parameter Space Exploration

"""
Visualize how MCMC explores the 4D parameter space by projecting 
onto 2D planes. Shows how guided learner finds target faster.
"""

function track_full_trajectory(trace, n_steps)
    trajectory = Vector{NTuple{4, Bool}}()
    
    for step in 0:n_steps
        state = (
            trace[:head_direction],
            trace[:subject_drop],
            trace[:adj_order],
            trace[:wh_fronting]
        )
        push!(trajectory, state)
        
        if step < n_steps
            trace = mh_step(trace)
        end
    end
    
    return trajectory
end

# Track trajectories
naive_trace_fresh, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), obs_map)
naive_traj = track_full_trajectory(naive_trace_fresh, 200)

nn_preds_fresh, _ = predict_nn(observed_input_vec)
guided_init = deepcopy(obs_map)
guided_init[:head_direction] = nn_preds_fresh.hd
guided_init[:subject_drop] = nn_preds_fresh.sd
guided_init[:adj_order] = nn_preds_fresh.ao
guided_init[:wh_fronting] = nn_preds_fresh.wh
guided_trace_fresh, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), guided_init)
guided_traj = track_full_trajectory(guided_trace_fresh, 200)

# Convert to numeric with jitter for visualization
function traj_to_coords(traj, param1_idx, param2_idx)
    jitter = 0.05
    x = [Float64(t[param1_idx]) + randn() * jitter for t in traj]
    y = [Float64(t[param2_idx]) + randn() * jitter for t in traj]
    return x, y
end

# Create 2x3 grid of parameter pair projections
param_pairs = [
    (1, 2, "Head Dir", "Subj Drop"),
    (1, 3, "Head Dir", "Adj Order"),
    (1, 4, "Head Dir", "Wh-Front"),
    (2, 3, "Subj Drop", "Adj Order"),
    (2, 4, "Subj Drop", "Wh-Front"),
    (3, 4, "Adj Order", "Wh-Front")
]

target_vec = (target_theta.hd, target_theta.sd, target_theta.ao, target_theta.wh)

p_space = plot(layout=(2, 3), size=(1400, 800),
               plot_title="Parameter Space Exploration (MCMC Trajectories)")

for (i, (p1, p2, n1, n2)) in enumerate(param_pairs)
    nx, ny = traj_to_coords(naive_traj, p1, p2)
    gx, gy = traj_to_coords(guided_traj, p1, p2)
    
    # Plot trajectories
    plot!(p_space[i], nx, ny,
          label="Naive", color=:red, alpha=0.4, linewidth=1,
          xlabel=n1, ylabel=n2,
          xlim=(-0.2, 1.2), ylim=(-0.2, 1.2))
    plot!(p_space[i], gx, gy,
          label="Guided", color=:blue, alpha=0.6, linewidth=1)
    
    # Mark start points
    scatter!(p_space[i], [nx[1]], [ny[1]], 
             color=:red, markersize=8, markershape=:circle, label="")
    scatter!(p_space[i], [gx[1]], [gy[1]],
             color=:blue, markersize=8, markershape=:circle, label="")
    
    # Mark target
    scatter!(p_space[i], [Float64(target_vec[p1])], [Float64(target_vec[p2])],
             color=:green, markersize=12, markershape=:star5, label="Target")
end

display(p_space)
```

```julia
## Cross-Language Generalization Analysis

"""
Test if the neural network generalizes to ALL 16 possible language types,
not just the one it was tested on.
"""

function evaluate_all_languages(n_trials_per_lang=10)
    results = []
    
    # All 16 combinations of 4 binary parameters
    for hd in [false, true]
        for sd in [false, true]
            for ao in [false, true]
                for wh in [false, true]
                    target = (hd=hd, sd=sd, ao=ao, wh=wh)
                    
                    naive_convs = Int[]
                    guided_convs = Int[]
                    nn_correct = Int[]
                    
                    for _ in 1:n_trials_per_lang
                        # Generate data from this language
                        constraints = Gen.choicemap()
                        constraints[:head_direction] = hd
                        constraints[:subject_drop] = sd
                        constraints[:adj_order] = ao
                        constraints[:wh_fronting] = wh
                        
                        tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), constraints)
                        sents = get_retval(tr)
                        input_vec = encode_corpus(sents)
                        
                        # Create observation map
                        test_obs = Gen.choicemap()
                        for i in 1:NUM_SENTS_TRAIN
                            test_obs[(:utt_type, i)] = tr[(:utt_type, i)]
                            test_obs[(:subj, i)] = tr[(:subj, i)]
                            test_obs[(:obj, i)] = tr[(:obj, i)]
                            test_obs[(:adj, i)] = tr[(:adj, i)]
                            if tr[(:utt_type, i)] == 1 || tr[(:utt_type, i)] == 3
                                test_obs[(:v_trans, i)] = tr[(:v_trans, i)]
                            else
                                test_obs[(:v_intr, i)] = tr[(:v_intr, i)]
                            end
                            for j in 1:MAX_SENT_LEN
                                test_obs[(:token, i, j)] = tr[(:token, i, j)]
                            end
                        end
                        
                        # Naive learner
                        naive_tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), test_obs)
                        naive_params, _ = track_parameters(naive_tr, 200)
                        _, _, n_conv = compute_convergence_metrics(naive_params, target)
                        push!(naive_convs, n_conv)
                        
                        # Guided learner
                        preds, _ = predict_nn(input_vec)
                        guided_map = deepcopy(test_obs)
                        guided_map[:head_direction] = preds.hd
                        guided_map[:subject_drop] = preds.sd
                        guided_map[:adj_order] = preds.ao
                        guided_map[:wh_fronting] = preds.wh
                        guided_tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), guided_map)
                        guided_params, _ = track_parameters(guided_tr, 200)
                        _, _, g_conv = compute_convergence_metrics(guided_params, target)
                        push!(guided_convs, g_conv)
                        
                        # NN accuracy
                        correct = (preds.hd == hd && preds.sd == sd && 
                                   preds.ao == ao && preds.wh == wh)
                        push!(nn_correct, correct ? 1 : 0)
                    end
                    
                    push!(results, (
                        target = target,
                        naive_mean = mean(naive_convs),
                        guided_mean = mean(guided_convs),
                        speedup = mean(naive_convs) / max(mean(guided_convs), 1),
                        nn_accuracy = mean(nn_correct)
                    ))
                end
            end
        end
    end
    
    return results
end

println("Evaluating all 16 language types (this may take a minute)...")
@time all_lang_results = evaluate_all_languages(5)

# Create heatmap of NN accuracy across language types
accuracy_matrix = zeros(4, 4)
speedup_matrix = zeros(4, 4)
labels_x = ["OV-NoDrop", "OV-Drop", "VO-NoDrop", "VO-Drop"]
labels_y = ["NA-InSitu", "NA-Front", "AN-InSitu", "AN-Front"]

for (i, r) in enumerate(all_lang_results)
    row = ((r.target.ao ? 1 : 0) * 2 + (r.target.wh ? 1 : 0)) + 1
    col = ((r.target.hd ? 1 : 0) * 2 + (r.target.sd ? 1 : 0)) + 1
    accuracy_matrix[row, col] = r.nn_accuracy
    speedup_matrix[row, col] = min(r.speedup, 200)  # Cap for visualization
end

p_gen = plot(layout=(1, 2), size=(1200, 500))

heatmap!(p_gen[1], labels_x, labels_y, accuracy_matrix,
         title="NN Prediction Accuracy\nAcross All Language Types",
         color=:RdYlGn, clims=(0, 1),
         xlabel="Head Dir + Subj Drop", ylabel="Adj Order + Wh-Front")

heatmap!(p_gen[2], labels_x, labels_y, speedup_matrix,
         title="Speedup (Guided vs Naive)\nAcross All Language Types",
         color=:Blues,
         xlabel="Head Dir + Subj Drop", ylabel="Adj Order + Wh-Front")

display(p_gen)

# Summary statistics
println("\n" * "=" ^ 60)
println("CROSS-LANGUAGE GENERALIZATION SUMMARY")
println("=" ^ 60)
println("Mean NN accuracy across all languages: $(round(mean([r.nn_accuracy for r in all_lang_results]), digits=3))")
println("Mean speedup across all languages: $(round(mean([r.speedup for r in all_lang_results]), digits=2))x")
println("Languages where NN achieves >80% accuracy: $(sum([r.nn_accuracy > 0.8 for r in all_lang_results]))/16")
```

```julia
## Sample Efficiency Analysis

"""
How does performance change with different corpus sizes?
This tests the poverty of the stimulus hypothesis - 
can the guided learner succeed with fewer examples?
"""

function evaluate_sample_efficiency(corpus_sizes=[2, 5, 10, 20, 50], n_trials=10)
    results = Dict{Int, NamedTuple}()
    
    for n_sents in corpus_sizes
        println("Testing corpus size: $n_sents sentences...")
        
        naive_convs = Int[]
        guided_convs = Int[]
        
        for _ in 1:n_trials
            # Generate target language data
            constraints = deepcopy(target_constraints)
            tr, _ = generate(corpus_model, (n_sents,), constraints)
            sents = get_retval(tr)
            
            # Need to handle variable input size for NN
            # Pad or truncate to NUM_SENTS_TRAIN
            padded_sents = Vector{Vector{String}}()
            for i in 1:NUM_SENTS_TRAIN
                if i <= length(sents)
                    push!(padded_sents, sents[i])
                else
                    push!(padded_sents, ["PAD"])
                end
            end
            input_vec = encode_corpus(padded_sents)
            
            # Create observation map (only for sentences we have)
            test_obs = Gen.choicemap()
            for i in 1:min(n_sents, NUM_SENTS_TRAIN)
                test_obs[(:utt_type, i)] = tr[(:utt_type, i)]
                test_obs[(:subj, i)] = tr[(:subj, i)]
                test_obs[(:obj, i)] = tr[(:obj, i)]
                test_obs[(:adj, i)] = tr[(:adj, i)]
                if tr[(:utt_type, i)] == 1 || tr[(:utt_type, i)] == 3
                    test_obs[(:v_trans, i)] = tr[(:v_trans, i)]
                else
                    test_obs[(:v_intr, i)] = tr[(:v_intr, i)]
                end
                for j in 1:MAX_SENT_LEN
                    test_obs[(:token, i, j)] = tr[(:token, i, j)]
                end
            end
            
            # Naive learner
            naive_tr, _ = generate(corpus_model, (n_sents,), test_obs)
            naive_params, _ = track_parameters(naive_tr, 300)
            _, _, n_conv = compute_convergence_metrics(naive_params, target_theta)
            push!(naive_convs, n_conv)
            
            # Guided learner
            preds, _ = predict_nn(input_vec)
            guided_map = deepcopy(test_obs)
            guided_map[:head_direction] = preds.hd
            guided_map[:subject_drop] = preds.sd
            guided_map[:adj_order] = preds.ao
            guided_map[:wh_fronting] = preds.wh
            guided_tr, _ = generate(corpus_model, (n_sents,), guided_map)
            guided_params, _ = track_parameters(guided_tr, 300)
            _, _, g_conv = compute_convergence_metrics(guided_params, target_theta)
            push!(guided_convs, g_conv)
        end
        
        results[n_sents] = (
            naive_mean = mean(naive_convs),
            naive_std = std(naive_convs),
            guided_mean = mean(guided_convs),
            guided_std = std(guided_convs)
        )
    end
    
    return results
end

println("\nRunning sample efficiency analysis...")
@time sample_results = evaluate_sample_efficiency([2, 5, 10, 15, 20], 8)

# Visualization
corpus_sizes = sort(collect(keys(sample_results)))
naive_means = [sample_results[k].naive_mean for k in corpus_sizes]
naive_stds = [sample_results[k].naive_std for k in corpus_sizes]
guided_means = [sample_results[k].guided_mean for k in corpus_sizes]
guided_stds = [sample_results[k].guided_std for k in corpus_sizes]

p_sample = plot(size=(900, 500),
                title="Sample Efficiency: Convergence vs Corpus Size",
                xlabel="Number of Training Sentences",
                ylabel="Steps to Convergence",
                legend=:topright)

plot!(p_sample, corpus_sizes, naive_means,
      ribbon=naive_stds,
      fillalpha=0.2,
      label="Naive Learner",
      color=:red,
      linewidth=3,
      marker=:circle,
      markersize=8)

plot!(p_sample, corpus_sizes, guided_means,
      ribbon=guided_stds,
      fillalpha=0.2,
      label="Guided Learner",
      color=:blue,
      linewidth=3,
      marker=:square,
      markersize=8)

display(p_sample)

# Compute and display advantage ratio
println("\n" * "="^50)
println("SAMPLE EFFICIENCY RESULTS")
println("="^50)
for k in corpus_sizes
    r = sample_results[k]
    advantage = r.naive_mean / max(r.guided_mean, 1)
    println("Corpus size $k: Naive=$(round(r.naive_mean, digits=1))±$(round(r.naive_std, digits=1)), " *
            "Guided=$(round(r.guided_mean, digits=1))±$(round(r.guided_std, digits=1)), " *
            "Speedup=$(round(advantage, digits=1))x")
end
```

```julia
## Summary Dashboard

"""
A comprehensive summary visualization showing all key results.
"""
function create_summary_dashboard(naive_convs, guided_convs, nn_probs, nn_preds, target)
    
    p = plot(layout=@layout([a b; c{0.3h} d{0.3h}]), 
             size=(1400, 900),
             plot_title="Universal Grammar Learning: Summary Dashboard")
    
    # Panel 1: Convergence Distribution
    boxplot!(p[1], 
             ["Naive\nLearner", "Guided\nLearner"],
             [naive_convs, guided_convs],
             title="Convergence Speed",
             ylabel="MCMC Steps to Converge",
             legend=false,
             color=[:indianred :steelblue],
             fillalpha=0.7,
             linewidth=2)
    
    # Add individual points
    scatter!(p[1], ones(length(naive_convs)) .+ randn(length(naive_convs)) .* 0.05,
             naive_convs, color=:red, alpha=0.5, label="", markersize=4)
    scatter!(p[1], 2 .* ones(length(guided_convs)) .+ randn(length(guided_convs)) .* 0.05,
             guided_convs, color=:blue, alpha=0.5, label="", markersize=4)
    
    # Panel 2: Effect Size Visualization
    speedups = naive_convs ./ max.(guided_convs, 1)
    histogram!(p[2], speedups,
               title="Speedup Distribution (Naive/Guided)",
               xlabel="Speedup Factor",
               ylabel="Frequency",
               color=:purple,
               alpha=0.7,
               legend=false,
               bins=15)
    vline!(p[2], [mean(speedups)], color=:red, linewidth=3, 
           linestyle=:dash, label="Mean: $(round(mean(speedups), digits=1))x")
    vline!(p[2], [median(speedups)], color=:orange, linewidth=2,
           linestyle=:dot, label="Median: $(round(median(speedups), digits=1))x")
    
    # Panel 3: NN Confidence
    param_labels = ["Head\nDirection", "Subject\nDrop", "Adjective\nOrder", "Wh-\nFronting"]
    bar!(p[3], param_labels, collect(nn_probs),
         title="Neural Network Confidence",
         ylabel="P(param=1)",
         legend=false,
         color=[:steelblue, :steelblue, :steelblue, :steelblue],
         fillalpha=0.7,
         ylim=(0, 1.1))
    hline!(p[3], [0.5], linestyle=:dash, color=:gray, linewidth=2)
    
    # Add target markers
    target_vals = [target.hd, target.sd, target.ao, target.wh]
    for (i, tv) in enumerate(target_vals)
        scatter!(p[3], [i], [Float64(tv)], 
                 color=:green, markersize=12, markershape=:star5, label="")
    end
    
    # Panel 4: Key Metrics (using simple text, no box drawing)
    plot!(p[4], axis=false, grid=false, ticks=false, 
          xlim=(0,1), ylim=(0,1))
    
    # Calculate metrics
    speedup_mean = round(mean(speedups), digits=1)
    nn_acc = round(mean([nn_preds.hd == target.hd, 
                         nn_preds.sd == target.sd,
                         nn_preds.ao == target.ao,
                         nn_preds.wh == target.wh]) * 100, digits=0)
    
    # Add text annotations instead of a text box
    annotate!(p[4], 0.5, 0.85, text("KEY PERFORMANCE METRICS", :center, 14, :bold))
    annotate!(p[4], 0.5, 0.73, text("─────────────────────────────", :center, 10))
    
    annotate!(p[4], 0.25, 0.60, text("Naive Learner:", :left, 11, :bold))
    annotate!(p[4], 0.30, 0.50, text("Mean: $(round(mean(naive_convs), digits=1)) steps", :left, 10))
    annotate!(p[4], 0.30, 0.42, text("Std:  $(round(std(naive_convs), digits=1)) steps", :left, 10))
    
    annotate!(p[4], 0.55, 0.60, text("Guided Learner:", :left, 11, :bold))
    annotate!(p[4], 0.60, 0.50, text("Mean: $(round(mean(guided_convs), digits=1)) steps", :left, 10))
    annotate!(p[4], 0.60, 0.42, text("Std:  $(round(std(guided_convs), digits=1)) steps", :left, 10))
    
    annotate!(p[4], 0.5, 0.25, text("─────────────────────────────", :center, 10))
    annotate!(p[4], 0.5, 0.15, text("Mean Speedup: $(speedup_mean)x  |  NN Accuracy: $(nn_acc)%", :center, 12, :bold))
    
    return p
end

# Generate data for dashboard
println("Generating dashboard data...")
dashboard_trials = 30
dashboard_results = [run_comparison_trial() for _ in 1:dashboard_trials]
dash_naive = [r[1] for r in dashboard_results]
dash_guided = [r[2] for r in dashboard_results]

p_dashboard = create_summary_dashboard(dash_naive, dash_guided, nn_probs, nn_preds, target_theta)
display(p_dashboard)

# Statistical test
using Statistics
println("\n" * "=" ^ 60)
println("STATISTICAL ANALYSIS")
println("=" ^ 60)
effect_size = (mean(dash_naive) - mean(dash_guided)) / sqrt((std(dash_naive)^2 + std(dash_guided)^2) / 2)
println("Cohen's d effect size: $(round(effect_size, digits=2))")
println("  (>0.8 is considered large)")
println("\n95% CI for speedup: [$(round(quantile(dash_naive ./ max.(dash_guided, 1), 0.025), digits=1)), $(round(quantile(dash_naive ./ max.(dash_guided, 1), 0.975), digits=1))]")
```

```julia
## Noise Sensitivity Analysis

"""
How robust is the guided learner to different noise levels?
"""

function evaluate_noise_sensitivity(noise_levels=[0.0, 0.05, 0.1, 0.15, 0.2, 0.25], n_trials=8)
    results = Dict{Float64, NamedTuple}()
    
    for noise in noise_levels
        println("Testing noise level: $noise...")
        
        naive_convs = Int[]
        guided_convs = Int[]
        
        for _ in 1:n_trials
            # Would need to regenerate model with different NOISE_PROB
            # For now, we simulate by adding perturbation to observations
            
            tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), target_constraints)
            sents = get_retval(tr)
            input_vec = encode_corpus(sents)
            
            # Add noise to input vector (simulating noisier observations)
            noisy_input = input_vec .+ randn(length(input_vec)) .* noise
            noisy_input = clamp.(noisy_input, 0.0, 1.0)
            
            # Naive (uses clean obs_map since it's doing inference)
            naive_tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), obs_map)
            naive_params, _ = track_parameters(naive_tr, 200)
            _, _, n_conv = compute_convergence_metrics(naive_params, target_theta)
            push!(naive_convs, n_conv)
            
            # Guided (uses potentially noisy NN prediction)
            preds, _ = predict_nn(noisy_input)
            guided_map = deepcopy(obs_map)
            guided_map[:head_direction] = preds.hd
            guided_map[:subject_drop] = preds.sd
            guided_map[:adj_order] = preds.ao
            guided_map[:wh_fronting] = preds.wh
            guided_tr, _ = generate(corpus_model, (NUM_SENTS_TRAIN,), guided_map)
            guided_params, _ = track_parameters(guided_tr, 200)
            _, _, g_conv = compute_convergence_metrics(guided_params, target_theta)
            push!(guided_convs, g_conv)
        end
        
        results[noise] = (
            naive_mean = mean(naive_convs),
            guided_mean = mean(guided_convs),
            guided_std = std(guided_convs),
            speedup = mean(naive_convs) / max(mean(guided_convs), 1)
        )
    end
    
    return results
end

println("\nRunning noise sensitivity analysis...")
@time noise_results = evaluate_noise_sensitivity()

# Visualization
noise_levels = sort(collect(keys(noise_results)))
guided_means = [noise_results[k].guided_mean for k in noise_levels]
guided_stds = [noise_results[k].guided_std for k in noise_levels]
speedups = [noise_results[k].speedup for k in noise_levels]

p_noise = plot(layout=(1, 2), size=(1100, 400))

plot!(p_noise[1], noise_levels, guided_means,
      ribbon=guided_stds,
      fillalpha=0.3,
      title="Guided Learner Robustness to Noise",
      xlabel="Input Noise Level (σ)",
      ylabel="Steps to Convergence",
      label="Guided Learner",
      color=:blue,
      linewidth=3,
      marker=:circle)

plot!(p_noise[2], noise_levels, speedups,
      title="Speedup vs Noise Level",
      xlabel="Input Noise Level (σ)",
      ylabel="Speedup (Naive/Guided)",
      label="Speedup",
      color=:purple,
      linewidth=3,
      marker=:diamond,
      markersize=8)

display(p_noise)
```
